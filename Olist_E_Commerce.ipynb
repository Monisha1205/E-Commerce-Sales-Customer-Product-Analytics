# Step 1 — Create the main analysis dataset
# First, combine the important tables:

```python
# ==========================================
# BUSINESS ANALYSIS — CREATE SALES DATASET
# ==========================================

sales = order_items.merge(
    orders[['order_id', 'order_purchase_timestamp', 'customer_id']],
    on='order_id',
    how='left'
)

sales = sales.merge(
    customers[['customer_id', 'customer_unique_id', 'customer_state']],
    on='customer_id',
    how='left'
)

sales = sales.merge(
    products[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

sales = sales.merge(
    category_translation,
    on='product_category_name',
    how='left'
)

# Create purchase month
sales['purchase_month'] = (
    sales['order_purchase_timestamp']
    .dt.to_period('M')
    .astype(str)
)

# Check the result
print("Sales dataset shape:", sales.shape)

sales.head()
```

```python
print("Sales dataset shape:", sales.shape)
sales.head()
```

```python
# ==========================================
# STEP 2 — MAIN KPI CALCULATION
# ==========================================

# 1. Total Revenue
total_revenue = sales['price'].sum()

# 2. Total Orders
total_orders = sales['order_id'].nunique()

# 3. Unique Customers
total_customers = sales['customer_unique_id'].nunique()

# 4. Average Order Value (AOV)
aov = total_revenue / total_orders

# 5. Average Review Score
average_review_score = reviews['review_score'].mean()

# 6. Average Delivery Time
average_delivery_days = orders['delivery_days'].mean()

# 7. Number of Late Orders
late_orders = (orders['delivery_performance'] == 'Late').sum()

# Display KPIs
print("========== OLIST E-COMMERCE KPIs ==========")
print(f"Total Revenue        : R$ {total_revenue:,.2f}")
print(f"Total Orders         : {total_orders:,}")
print(f"Unique Customers     : {total_customers:,}")
print(f"Average Order Value  : R$ {aov:,.2f}")
print(f"Average Review Score : {average_review_score:.2f}")
print(f"Average Delivery Time: {average_delivery_days:.2f} days")
print(f"Late Orders          : {late_orders:,}")
```

| KPI | Meaning |
|---|---|
| **Total Revenue** | Total product sales revenue |
| **Total Orders** | Number of unique orders |
| **Unique Customers** | Number of distinct customers |
| **AOV** | Average amount spent per order |
| **Average Review Score** | Overall customer rating |
| **Average Delivery Time** | Average days taken for delivery |
| **Late Orders** | Orders delivered after the estimated date |

```python
# ==========================================
# STEP 3 — MONTHLY REVENUE TREND
# ==========================================

# Calculate revenue for each month
monthly_revenue = (
    sales.groupby('purchase_month')['price']
    .sum()
    .reset_index()
)

# Rename columns
monthly_revenue.columns = ['Month', 'Revenue']

# Calculate month-over-month revenue growth
monthly_revenue['Revenue_Growth_%'] = (
    monthly_revenue['Revenue'].pct_change() * 100
)

# Display results
monthly_revenue.head(12)
```

# Positive vs Negative reviews

```python
# ==========================================
# POSITIVE / NEUTRAL / NEGATIVE REVIEWS
# ==========================================

def classify_review(score):
    if score >= 4:
        return 'Positive'
    elif score == 3:
        return 'Neutral'
    else:
        return 'Negative'

reviews['Review_Category'] = reviews['review_score'].apply(classify_review)

review_category_summary = (
    reviews['Review_Category']
    .value_counts()
    .reset_index()
)

review_category_summary.columns = ['Review_Category', 'Review_Count']

review_category_summary['Percentage'] = (
    review_category_summary['Review_Count']
    / review_category_summary['Review_Count'].sum()
    * 100
).round(2)

display(review_category_summary)
```

# Prepare delivery and review data

```python
# ==========================================
# STEP 10 — DELIVERY TIME VS CUSTOMER REVIEWS
# ==========================================

# Calculate one average review score per order
order_reviews = (
    reviews.groupby('order_id')
    .agg(Review_Score=('review_score', 'mean'))
    .reset_index()
)

# Select delivery details
delivery_data = orders[
    ['order_id', 'order_status', 'delivery_performance', 'delivery_days']
].merge(
    order_reviews,
    on='order_id',
    how='left'
)

# Keep delivered orders with valid delivery times and reviews
delivery_data = delivery_data[
    (delivery_data['order_status'] == 'delivered') &
    (delivery_data['delivery_days'].notna()) &
    (delivery_data['Review_Score'].notna())
].copy()

print("Orders included in analysis:", len(delivery_data))
display(delivery_data.head())
```

# Compare average review scores

```python
# Average rating by delivery performance
delivery_review_summary = (
    delivery_data.groupby('delivery_performance')
    .agg(
        Order_Count=('order_id', 'nunique'),
        Average_Review_Score=('Review_Score', 'mean'),
        Average_Delivery_Days=('delivery_days', 'mean')
    )
    .reset_index()
)

delivery_review_summary['Average_Review_Score'] = (
    delivery_review_summary['Average_Review_Score'].round(2)
)

delivery_review_summary['Average_Delivery_Days'] = (
    delivery_review_summary['Average_Delivery_Days'].round(2)
)

display(delivery_review_summary)
```

# Create the comparison chart

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=delivery_review_summary,
    x='delivery_performance',
    y='Average_Review_Score'
)

plt.title('Delivery Performance vs Average Customer Review')
plt.xlabel('Delivery Performance')
plt.ylabel('Average Review Score (1–5)')
plt.ylim(0, 5)
plt.tight_layout()
plt.show()
```

# Calculate the rating difference

```python
rating_comparison = delivery_review_summary.set_index(
    'delivery_performance'
)['Average_Review_Score']

print("Average ratings by delivery performance:")
print(rating_comparison)

if 'Late' in rating_comparison.index:
    late_rating = rating_comparison['Late']
    other_ratings = rating_comparison[
        rating_comparison.index != 'Late'
    ]

    if not other_ratings.empty:
        on_time_rating = other_ratings.iloc[0]

        print(f"\nLate delivery rating: {late_rating:.2f}")
        print(f"Comparison group rating: {on_time_rating:.2f}")
        print(
            f"Rating difference: "
            f"{on_time_rating - late_rating:.2f} points"
        )
```
